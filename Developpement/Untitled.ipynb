```python
cat_col = ["Texture_sol","Couleur_sol","Type_irrigation","Niveau_Engrais","Temperature_Actuelle"]
num_col = ["Hauteur_Eau"]
y_col = "Densite_Tilapia_Are"
```

```python
sns.displot(df,x=y_col,kde=True)
plt.title(f"Displot {y_col}")
plt.savefig(f"./Images/Denite_tilapia/Displot {y_col}")
```

```python
for i in cat_col:
    
    #plt.savefig(f"Violin Plot de {i} par rapport au Densite_Tilapia_Are")
    #sns.displot(df[i])
    #sns.countplot(df,x=i,hue="Densite_Tilapia_Are")
    #sns.boxplot(data=df, x=i, y="Densite_Tilapia_Are")
    sns.violinplot(data=df, x=i, y="Densite_Tilapia_Are")
    #sns.stripplot(data=df, x=i, y="Densite_Tilapia_Are")
    #sns.displot(df,x="Densite_Tilapia_Are",hue=i,multiple="stack",bins=20)
    plt.title(f"Violin Plot {i}")
    plt.savefig(f"./Images/Violin Plot {i}")
    plt.close()
    
    
    
```

```python
dist = [4,8,14,20]
labels = ["5-8cm","9-14cm","15-20cm"]
for i in num_col:
    
    
    #sns.violinplot(data=df, x=i, y="Densite_Tilapia_Are")
    #sns.stripplot(data=df, x=i, y="Densite_Tilapia_Are")
    #sns.displot(df,x="Densite_Tilapia_Are",hue=i,multiple="stack",bins=20)
    #sns.histplot(df,x=i)
    temp_df = df[["Densite_Tilapia_Are"]].copy()
    temp_df[i] = pd.cut(df[i],bins=dist,labels=labels)
    sns.violinplot(data=temp_df, x=i, y="Densite_Tilapia_Are")
    plt.title(f"Violin Plot {i}")

    #plt.show()
    plt.savefig(f"./Images/ViolinPlot {i}")
    plt.close()
    
    
    
    
```
