```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
place = [[3,10,15,47,52,65],
         [2,9,17,48,53,66],
         [1,7,13,41,46,61],
         [1,8,16,39,42,55],
         [1,7,14,36,39,51],
         [1,7,14,47,35,50],
         [1,6,13,50,38,54],
         [1,7,14,50,38,54],
         [1,7,14,48,35,52]
        ]
```

```python
df = pd.read_excel("./Production par région 2015-2023.xlsx",sheet_name=None)
```

```python
place_index = [
    [2, 9, 14, 46, 51, 64],
    [1, 8, 16, 47, 52, 65],
    [0, 6, 12, 40, 45, 60],
    [0, 7, 15, 38, 41, 54],
    [0, 6, 13, 35, 38, 50],
    [0, 6, 13, 46, 34, 49],
    [0, 5, 12, 49, 37, 53],
    [0, 6, 13, 49, 37, 53],
    [0, 6, 13, 47, 34, 51]
]

label = ["Industrielle Maritime","Artisanale Maritime","Traditionnelle Maritime","Aquaculture Marine","Continentale","Aquaculture Eau Douce"]
```

```python
Tdf = {}

for l in label:
    Tdf[l] = []
Tdf["annee"] = []
i = 0
for k,v in df.items():
    temp_df = v.iloc[place_index[i],1:]
    temp_df = temp_df.apply(pd.to_numeric).sum(axis=1)
    for j in range(len(label)):
        Tdf[label[j]].append(temp_df.iloc[j])
    Tdf["annee"].append(k)
    i += 1
```

```python
Tdf = {}

Tdf["type"] = []
Tdf["annee"] = []
Tdf["valeur"] = []
i = 0
for k,v in df.items():
    temp_df = v.iloc[place_index[i],1:]
    temp_df = temp_df.apply(pd.to_numeric).sum(axis=1)
    for j in range(len(label)):
        Tdf["type"].append(label[j])
        Tdf["annee"].append(k)
        Tdf["valeur"].append(temp_df.iloc[j])
    i += 1
```

```python
Tdf = pd.DataFrame(Tdf)
```

```python
Tdf = Tdf.round(decimals=2)
```

```python
sns.lineplot(data=Tdf,x="annee",y="valeur",hue="type")
```
